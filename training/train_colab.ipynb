# Probe Detector -- Training on Google Colab

**Training only.** This notebook builds the dataset and trains every candidate
checkpoint, then stops -- it does not evaluate, compare, or pick a winner.
That analysis lives in a separate notebook, `training/analyze_results.ipynb`,
which reads the checkpoints this notebook produces and can be re-run any
number of times (including locally, on CPU, no GPU needed) without
retraining. Keeping the two apart avoids the repeated confusion of mixing a
long GPU training run with quick, iterative changes to evaluation code.

Trains 3 checkpoints: YOLOv8n, YOLOv8s, YOLOv8m, with otherwise identical
hyperparameters (model-size comparison, not a hyperparameter sweep -- see
`report/dataset_analysis.tex` for why). A horizontal-flip augmentation
ablation (motivated by the EDA's left-skewed probe position) was considered
-- training each variant a second time with `fliplr=0` -- but dropped for
time: doubling every training run wasn't worth it once the n/s/m comparison
alone turned out to be decisive. Noted as future work in the report rather
than left half-done here.

Each checkpoint and the built dataset are copied to Drive as soon as they're
ready, so a session disconnect doesn't lose completed work, and
`analyze_results.ipynb` doesn't need to rebuild the dataset (a LaMa
dependency it otherwise wouldn't need at all).

## 3. Get the project onto local VM disk

Reading thousands of small JPEGs directly from a mounted Drive is slow
(network I/O). This cell copies everything to `/content` (fast local disk,
but **not persistent** across sessions -- re-run every session).

Supports two ways of having gotten the project onto Drive:
- **A single `Probe_detector.zip`** uploaded to `MyDrive/Probe_detector.zip`
  (the practical option if you don't have Google Drive for Desktop --
  uploading one zip through the browser is far more reliable than
  drag-and-dropping hundreds of individual image files). Extracted directly
  here.
- **An already-extracted `MyDrive/Probe_detector/` folder** (e.g. synced via
  Google Drive for Desktop). Copied as-is.

The zip is checked first. This cell also defines `sync_to_drive()`, used
below to push the built dataset and each trained checkpoint back to Drive.

```python
import shutil
import zipfile
from datetime import datetime
from pathlib import Path

LOCAL_PROJECT_DIR = Path('/content/Probe_detector')
DRIVE_ZIP_PATH = Path('/content/drive/MyDrive/Probe_detector.zip')

if LOCAL_PROJECT_DIR.exists():
    shutil.rmtree(LOCAL_PROJECT_DIR)

if DRIVE_ZIP_PATH.exists():
    stat = DRIVE_ZIP_PATH.stat()
    print(f'Found zip on Drive: {DRIVE_ZIP_PATH}')
    print(f'  size: {stat.st_size:,} bytes')
    print(f'  modified: {datetime.fromtimestamp(stat.st_mtime)}')
    print('  >>> compare these against the zip file you uploaded (right-click it in the Drive web UI > File information) <<<')
    LOCAL_PROJECT_DIR.mkdir(parents=True)
    with zipfile.ZipFile(DRIVE_ZIP_PATH) as zf:
        zf.extractall(LOCAL_PROJECT_DIR)
    print(f'Extracted {DRIVE_ZIP_PATH} -> {LOCAL_PROJECT_DIR}')
elif Path(DRIVE_PROJECT_DIR).exists():
    shutil.copytree(DRIVE_PROJECT_DIR, LOCAL_PROJECT_DIR,
                     ignore=shutil.ignore_patterns('yolo_dataset', 'runs', '.venv', '.venv-lama', '.git'))
    print(f'Copied {DRIVE_PROJECT_DIR} -> {LOCAL_PROJECT_DIR}')
else:
    raise FileNotFoundError(
        f'Neither {DRIVE_ZIP_PATH} nor {DRIVE_PROJECT_DIR} found on Drive -- upload the project first.'
    )

%cd {LOCAL_PROJECT_DIR}

def sync_to_drive(local_relpath: str):
    """Copy a local file or folder (relative to the project root) to the
    same relative path on Drive. Safe to re-run -- overwrites/adds, never
    deletes."""
    src = LOCAL_PROJECT_DIR / local_relpath
    dst = Path(DRIVE_PROJECT_DIR) / local_relpath
    if src.is_dir():
        shutil.copytree(src, dst, dirs_exist_ok=True)
    else:
        dst.parent.mkdir(parents=True, exist_ok=True)
        shutil.copy(src, dst)
    print(f'Synced {local_relpath} -> Drive')
```

## 5. Train every candidate checkpoint

YOLOv8 n/s/m, with otherwise identical hyperparameters. Each checkpoint is
copied to `Drive/weights/<name>/best.pt` as soon as it finishes, so a
session disconnect only costs the run in progress, not everything before it.

```python
from training.train import train

VARIANTS = ['yolov8n.pt', 'yolov8s.pt', 'yolov8m.pt']
EPOCHS = 100
IMGSZ = 640
BATCH = 16

trained = {}
for variant in VARIANTS:
    name = variant.replace('.pt', '')
    print(f'\n=== Training {name} ===')
    best_path = train(
        data_yaml=Path('yolo_dataset/data.yaml'),
        model=variant,
        epochs=EPOCHS,
        imgsz=IMGSZ,
        batch=BATCH,
        patience=20,
        project=Path('runs/train'),
        name=name,
        seed=42,
        device='0',  # first GPU
    )
    trained[name] = best_path
    print(f'{name} -> {best_path}')

    local_weights_dir = Path('weights') / name
    local_weights_dir.mkdir(parents=True, exist_ok=True)
    shutil.copy(best_path, local_weights_dir / 'best.pt')
    sync_to_drive(f'weights/{name}')

print('\nAll candidates trained:', list(trained.keys()))
print('Next: run training/analyze_results.ipynb (or analyze_results.py locally) to evaluate, compare, and pick a winner.')
```
